## 🎯 Step 1 — Create the dataset
We'll have 10 students and 5 subjects:

````python
import numpy as np

students = np.array([
    "Student_1", "Student_2", "Student_3", "Student_4", "Student_5",
    "Student_6", "Student_7", "Student_8", "Student_9", "Student_10"
])

marks = np.array([
    [78, 65, 82, 71, 88],
    [92, 88, 79, 85, 90],
    [56, 61, 58, 64, 60],
    [85, 91, 89, 87, 93],
    [45, 52, 48, 55, 50],
    [73, 69, 75, 80, 77],
    [95, 90, 94, 92, 96],
    [62, 70, 65, 68, 72],
    [81, 78, 84, 76, 80],
    [50, 45, 55, 49, 58]
])
````

Here:

````python
marks.shape
````

## 🎯 Step 2 — Your task

Without running anything first, write code to calculate:

1. Total marks of each student
2. Average marks of each student

````python
# Total marks of each student

total_marks = marks.sum(axis=1)

print(f'Total marks of each student is: {total_marks}')
````

````python
# Average marks of each student

average_marks = marks.mean(axis=1)

print(f'Average marks of each student is: {average_marks}')
````

## 🎯 Step 3 — Find the top student

Now use what you learned with argmax().

We want to find:

1. The index of the student with the highest average.
2. The student's name.
3. Their average marks.

````python
# The index of the student with the highest average.

top_index = average_marks.argmax()
top_student = students[top_index]
top_average = average_marks[top_index]

print(f'The index of the student with the highest average is: {top_index}')
print(f"The student's name: {top_student}")
print(f'Their average marks: {top_average}')
````

## 🎯 Step 4 — Rank all students

Now let's use the argsort() you just learned.

We want students ordered from highest average → lowest average.

Create:

ranking = ...

````python
# ranking = average_marks.argsort()[::-1]
ranking = average_marks.argsort(descending=True)

ranking
````

## 🎯 Step 5 — Subject-wise analysis

Now let's switch from student-wise to subject-wise.

Remember:
```python
axis=0
```
means we operate down the rows, giving one result for each column.

Calculate:
```text
subject_average = ...
subject_max = ...
subject_min = ...
```

````python
subject_average = marks.mean(axis=0)
subject_max = marks.max(axis=0)
subject_min = marks.min(axis=0)

print(f'Average marks in each subject is: {subject_average}')
print(f'Maximum marks in each subject is: {subject_max}')
print(f'Minimum marks in each subject is: {subject_min}')
````

## 🎯 Step 6 — Performance categories

Now let's use one of your important NumPy skills: np.select().

We'll categorize each student's average:
```text
90+       → Excellent
75–89.99  → Good
60–74.99  → Average
<60       → Poor
```
Create:

````python
conditions = [
    average_marks >= 90,
    average_marks >= 75,
    average_marks >= 60
]

choices = [
    "Excellent",
    "Good",
    "Average",
]

performance = np.select(
    conditions,
    choices,
    default="Poor"
)

print(f'Performance of each student: {performance}')
````

## 🎯 Step 7 — Find students above the overall average

Now let's combine a few things we've learned.

Calculate the overall average of all marks:

overall_average = ...

Then find all students whose individual average is greater than the overall average:

above_average = ...

````python
overall_average = marks.mean()
print(overall_average)

above_average = students[average_marks > overall_average]
print(above_average)
````

## 🎯 Step 8 — Count performance categories

Now use np.unique() with return_counts=True.

````python
categories, counts = np.unique(performance, return_counts=True)

print(f'categories: {categories}\ncounts: {counts}')
````

Remember, np.unique() sorts the categories, so they appear alphabetically.
