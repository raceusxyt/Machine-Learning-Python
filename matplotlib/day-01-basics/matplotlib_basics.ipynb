## 1. What is Matplotlib?

Matplotlib is a Python library used to create visualizations and graphs from data.

For example, suppose you have:
```python
expenses = [200, 350, 150, 500, 400]
```
Instead of only looking at the numbers, Matplotlib lets us visualize them:
```python
500 |             ●
400 |                 ●
350 |     ●
300 |
200 | ●
150 |         ●
    +--------------------
      Mon Tue Wed Thu Fri
```
This becomes especially useful when working with the Pandas DataFrames you've already learned.

## 2. Importing Matplotlib

The most commonly used part of Matplotlib is pyplot.
```python
import matplotlib.pyplot as plt
```
We usually give it the alias:

plt

So instead of writing:
```python
matplotlib.pyplot.plot()
```
we can simply write:
```python
plt.plot()
```

## 3. Your First Plot

Start with this in your notebook:

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y)

plt.show()
````

**What happens?**
```python
plt.plot(x, y) tells Matplotlib:
```
Plot the values of y against the values of x.

So the points are:
```python
(1, 10)
(2, 20)
(3, 15)
(4, 30)
(5, 25)
```
Matplotlib then connects these points with lines.

## 4. Understanding x and y

This is important.
```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y)
```
Think of it as:

x → horizontal axis
y → vertical axis

So:
```python
       y
       ↑
  30   |           ●
  25   |               ●
  20   |     ●
  15   |         ●
  10   | ●
       +----------------→ x
          1  2  3  4  5
```

## 5. What if we provide only one list?

Try:

````python
y = [10, 20, 15, 30, 25]

plt.plot(y)

plt.show()
````

Here we didn't provide x.

Matplotlib automatically generates:
```python
x = [0, 1, 2, 3, 4]
```
So it effectively plots:
```python
x = [0, 1, 2, 3, 4]
y = [10, 20, 15, 30, 25]
```
This is very useful when you simply want to visualize a sequence of values.

## 6. Adding a Title

Use:
```python
plt.title("Weekly Sales")

plt.show()
```
For example:

````python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y)

plt.title("Weekly Sales")

plt.show()
````

## 7. Adding Axis Labels

Use:
```python
plt.xlabel("Day")
plt.ylabel("Sales")
```
Complete example:

````python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y)

plt.title("Weekly Sales")
plt.xlabel("Day")
plt.ylabel("Sales")

plt.show()
````

Now your graph communicates much more information.

## 8. Using NumPy with Matplotlib

This is where your previous NumPy knowledge becomes useful.

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(1, 6)
y = np.array([10, 20, 15, 30, 25])

plt.plot(x, y)

plt.title("Weekly Sales")
plt.xlabel("Day")
plt.ylabel("Sales")

plt.show()
````

Here:
```python
np.arange(1, 6)
```
produces:
```python
[1 2 3 4 5]
```
So NumPy generates/manipulates the data, while Matplotlib visualizes it.

## 9. The Basic Matplotlib Workflow

For now, remember this basic pattern:

import matplotlib.pyplot as plt
```python
# Data
x = [...]
y = [...]

# Create plot
plt.plot(x, y)

# Customize
plt.title("...")
plt.xlabel("...")
plt.ylabel("...")

# Display
plt.show()
```
Later we'll learn much more powerful ways to customize and organize plots.
