## 1. Basic Line Plot
Start your notebook with:

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y)

plt.show()
````

```python
plt.plot(x, y) creates a line plot by connecting the points:
(1,10)
(2,20)
(3,15)
(4,30)
(5,25)
```
The important idea is:
* x → horizontal axis
* y → vertical axis

## 2. Line Plot with NumPy

Since you've already learned NumPy, you'll often encounter NumPy arrays instead of Python lists.

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(1, 6)
y = np.array([10, 20, 15, 30, 25])

plt.plot(x, y)

plt.show()
````

This works exactly the same way.

## 3. Adding a Title and Labels

````python
plt.plot(x, y)

plt.title("Weekly Sales")
plt.xlabel("Day")
plt.ylabel("Sales")

plt.show()
````

This is something you'll use constantly in data visualization.

## 4. Multiple Lines

One of the most useful things about line plots is comparing multiple datasets.
For example:

````python
days = [1, 2, 3, 4, 5]

sales_a = [10, 20, 15, 30, 25]
sales_b = [8, 15, 20, 25, 30]

plt.plot(days, sales_a)
plt.plot(days, sales_b)

plt.show()
````

You now have two lines on the same graph.

## 5. Labels and Legend

When you have multiple lines, you should tell the viewer what each line represents.

````python
plt.plot(days, sales_a, label="Product A")
plt.plot(days, sales_b, label="Product B")

plt.title("Product Sales")
plt.xlabel("Day")
plt.ylabel("Sales")

plt.legend()

plt.show()
````

label= gives a name to the line.
plt.legend() displays those names.

**Important:**
```python
label="Product A"
```
does not display the label by itself.

You need:
```python
plt.legend()
```
to actually show it.

## 6. Markers
Markers show the individual data points.

````python
plt.plot(x, y, marker="o")

plt.show()
````

"o" means a circular marker.

Some useful markers:
* "o"  → circle
* "s"  → square
* "^"  → triangle
* "*"  → star
* "x"  → x
* "+"  → plus

**Changing the Marker Size**\
You can control marker size with:

````python
plt.plot(
    x,
    y,
    marker="o",
    markersize=8
)

plt.show()
````

## 7. Line Styles
You can change how the line appears.\
**Solid line:**

````python
plt.plot(x, y, linestyle="-")

plt.show()
````

**Dashed line:**

````python
plt.plot(x, y, linestyle="--")

plt.show()
````

**Dotted line:**

````python
plt.plot(x, y, linestyle=":")

plt.show()
````

**Dash-dot:**

````python
plt.plot(x, y, linestyle="-.")

plt.show()
````

So:
* *-*   → solid
* --  → dashed
* :   → dotted
* -.  → dash-dot

## 8. Line Width
Use linewidth or its shorter form lw.

````python
plt.plot(x, y, linewidth=3)

plt.show()
````

or

````python
plt.plot(x, y, lw=3)

plt.show()
````

A larger value means a thicker line.

## 9. Line Color
Use color.

````python
plt.plot(x, y, color="red")

plt.show()
````

You can use common color names such as:
```python
"red"
"blue"
"green"
"black"
"orange"
"purple"
```
You can also use shorthand:
```python
"r"  # red
"b"  # blue
"g"  # green
"k"  # black
"y"  # yellow
"m"  # magenta
"c"  # cyan
```
For learning, I'd recommend using the full names because they're easier to remember.

**Changing Marker and Line Separately**
You don't have to use the same color for everything.\
For example:

````python
plt.plot(days,
         food,
         color="blue",
         marker="o",
         ms=8,
         markerfacecolor="red",
         markeredgecolor="black"
        )

plt.show()
````

Here:
* line → blue
* marker → red
* marker edge → black

## 10. Combining Everything
Now let's combine what we've learned.

````python
days = [1, 2, 3, 4, 5]

sales_a = [10, 20, 15, 30, 25]
sales_b = [8, 15, 20, 25, 30]

plt.plot(
    days,
    sales_a,
    color="blue",
    marker="o",
    linestyle="-",
    lw=2,
    label="Product A"
)

plt.plot(
    days,
    sales_b,
    color="red",
    marker="s",
    linestyle="--",
    lw=2,
    label="Product B"
)

plt.title("Product Sales")
plt.xlabel("Day")
plt.ylabel("Sales")

plt.legend()

plt.show()
````

This is an important pattern to understand:
```python
plt.plot(
    x,
    y,
    color=...,
    marker=...,
    linestyle=...,
    linewidth=...,
    label=...
)
```
You don't need to memorize every parameter today. The important thing is knowing what each one controls.

## 11. Grid
A grid can make a graph easier to read.

````python
plt.plot(x, y)

plt.grid()

plt.show()
````

You can also control transparency:

````python
plt.plot(x, y)

plt.grid(alpha=0.8)

plt.show()
````

We'll learn more customization later.

## 🧠 Final Day 2 Challenge
You're given monthly data:
| Months | Sales | Expenses |
| :--- | :--- | :--- |
| 1 | 120 | 80 |
| 2 | 150 | 90 |
| 3 | 140 | 100 |
| 4 | 180 | 110 |
| 5 | 200 | 130 |
| 6 | 220 | 140 |

Create a line plot with these requirements:
- 📈 Plot both sales and expenses
- Sales → circle marker + solid line
- Expenses → square marker + dashed line
- Give them appropriate labels
- Title: "Sales vs Expenses"
- X-axis: "Month"
- Y-axis: "Amount"
- Add a legend
- Add a grid

````python
import numpy as np
import matplotlib.pyplot as plt

months = np.arange(1, 7)

sales = np.array([120, 150, 140, 180, 200, 220])
expenses = np.array([80, 90, 100, 110, 130, 140])

plt.plot(
    months,
    sales,
    linestyle="-",
    marker="o",
    label="Sales"
)

plt.plot(
    months,
    expenses,
    linestyle="--",
    marker="s",
    label="Expenses"
)

plt.title("Sales vs Expenses")
plt.xlabel("Months")
plt.ylabel("Amount")
plt.legend()
plt.grid()

plt.show()
````
