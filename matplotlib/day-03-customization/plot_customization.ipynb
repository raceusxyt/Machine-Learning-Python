## 1. Figure Size
So far we've simply written:
```python
plt.plot(x, y)
plt.show()
```
We can control the size of the figure using:
```python
plt.figure(figsize=(10, 5))
```
For example:

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.figure(figsize=(10, 5))

plt.plot(x, y)

plt.show()
````

The figsize argument is:
(width, height)

and the unit is inches.

So:
```python
figsize=(10, 5)
```
means:
* Width  = 10 inches
* Height = 5 inches

## 2. Figure vs Axes
This is an important concept in Matplotlib.

Think of it like this:
```text
Figure
└── Axes
    ├── X-axis
    ├── Y-axis
    └── Plot
```
The Figure is the entire canvas/window.\
The Axes is the actual plotting area inside the figure.

You can create a figure explicitly:
```python
fig = plt.figure(figsize=(10, 5))
```
We'll work more with fig and ax later when we get to subplots.

For now, just remember:
* Figure = entire canvas
* Axes   = actual plotting area

## 3. Setting X-axis Limits
Suppose:
```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]
```
Normally Matplotlib automatically chooses the axis range.

You can control it with:
```python
plt.xlim(1, 5)
```
Example:

````python
plt.plot(x, y)

plt.xlim(1, 5)

plt.show()
````

This tells Matplotlib:
```text
Set the x-axis range from 1 to 5.
```

## 4. Setting Y-axis Limits
Similarly:
```python
plt.ylim(0, 35)
```
Example:

````python
plt.plot(x, y)

plt.xlim(1, 5)
plt.ylim(0, 35)

plt.show()
````

Now you explicitly control both axes.
```python
plt.xlim(min, max)
plt.ylim(min, max)
```

## 5. Customizing Ticks
Ticks are the marks/values shown along the axes.

For example:
```python
plt.xticks([1, 2, 3, 4, 5])
```
You can specify exactly which x-values should appear.

For y:
```python
plt.yticks([0, 10, 20, 30, 40])
```
Example:

````python
plt.plot(x, y)

plt.xticks([1, 2, 3, 4, 5])
plt.yticks([0, 10, 20, 30, 40])

plt.show()
````

## 6. Custom Tick Labels
This becomes particularly useful with categorical data.

Suppose:
```python
x = [1, 2, 3, 4, 5]
y = [20, 25, 22, 30, 35]
```
Instead of showing: 1  2  3  4  5

we can show: Jan Feb Mar Apr May

using:

````python
x = [1, 2, 3, 4, 5]
y = [20, 25, 22, 30, 35]

plt.plot(x, y)

plt.xticks(
    [1, 2, 3, 4, 5],
    ["Jan", "Feb", "Mar", "Apr", "May"]
)

plt.show()
````

This is especially useful when working with real-world data.

## 7. Font Size
We can customize the title:
```python
plt.title("Monthly Revenue", fontsize=16)
```
Axis labels:
```python
plt.xlabel("Month", fontsize=12)
plt.ylabel("Revenue", fontsize=12)
```
For example:

````python
plt.plot(x, y)

plt.title("Monthly Revenue", fontsize=16)
plt.xlabel("Month", fontsize=12)
plt.ylabel("Revenue", fontsize=12)

plt.show()
````

## 8. Font Weight
You can make text bold:
```python
plt.title(
    "Monthly Revenue",
    fontsize=16,
    fontweight="bold"
)
```
You can also use:
```python
fontweight="normal"
```
or:
```python
fontweight="bold"
```

````python
plt.plot(x, y)

plt.title("Monthly Revenue",
          fontsize=16,
          fontweight="bold"
         )
plt.xlabel("Month",
           fontsize=12,
           fontweight="bold"
          )
plt.ylabel("Revenue",
           fontsize=12,
           fontweight="bold"
          )

plt.show()
````

## 9. Grid Customization
You've already learned:
```python
plt.grid()
```
Now let's customize it.
```python
plt.grid(
    linestyle="--",
    alpha=0.5
)
```
Here:
* linestyle="--" → dashed grid
* alpha=0.5      → semi-transparent

You can also control the width:
```python
plt.grid(
    linestyle="--",
    linewidth=0.8,
    alpha=0.5
)
```

````python
plt.plot(x, y)

plt.title("Monthly Revenue",
          fontsize=16,
          fontweight="bold"
         )
plt.xlabel("Month",
           fontsize=12,
           fontweight="bold"
          )
plt.ylabel("Revenue",
           fontsize=12,
           fontweight="bold"
          )
plt.grid(
    linestyle="--",
    linewidth=1.5,
    alpha=0.9
)

plt.show()
````

## 10. Rotating Tick Labels
Suppose your x-axis contains long labels:
```python
months = [    "January",    "February",    "March",    "April",    "May"]
```
They might overlap.

You can rotate them:
```python
plt.xticks(rotation=45)
```
For example:

````python
months = [
    "January",
    "February",
    "March",
    "April",
    "May"
]

plt.plot(months, [20, 25, 22, 30, 35])

plt.xticks(rotation=45)

plt.show()
````

This is very common when working with dates or long categorical labels.

## 11. Annotations
Annotations allow us to point out an important value on the graph.

For example:

````python
x = [1, 2, 3, 4, 6]
y = [10, 20, 15, 30, 40]

plt.plot(x, y)

plt.annotate(
    "Highest",
    xy=(4, 30),
    xytext=(3, 35)
)

plt.show()
````

Here:
```python
xy=(4, 30)
```
is the point we're talking about.

And:
```python
xytext=(3, 35)
```
is where the text is placed.

This is extremely useful for highlighting:
- Highest value
- Lowest value
- Important event
- Outlier
- Maximum sales
- Minimum expense

**Adding an Arrow**\
We can make the annotation clearer with an arrow:

````python
x = [1, 2, 3, 4, 6]
y = [10, 20, 15, 30, 40]

plt.plot(x, y)

plt.annotate(
    "Highest",
    xy=(4, 30),
    xytext=(3, 35),
    arrowprops=dict(arrowstyle="->")
)

plt.show()
````

Now the text points toward the data point.

**Text on a Plot**\
Sometimes you don't need an arrow.

You can simply place text using:

````python
x = [1, 2, 3, 4, 6]
y = [10, 20, 15, 30, 40]

plt.plot(x, y)

plt.text(
    3,
    32,
    "Important point"
)

plt.show()
````

The syntax is:
```python
plt.text(x_position, y_position, "text")
```
Difference
```python
plt.text()
    ↓
Just places text

plt.annotate()
    ↓
Can identify a specific point
and optionally use an arrow
```

## 12. Spines
Spines are the borders around the plotting area.

A normal plot has four:
```text
Top spine
    ┌───────────────┐
Left│               │Right
    │     Plot      │
    └───────────────┘
       Bottom spine
```
You can access them using:
```python
plt.gca().spines
```
gca() means: Get Current Axes

For example:

````python
plt.plot(x, y)

ax = plt.gca()

ax.spines

plt.show()
````

But usually we want to customize a particular spine.

**Hide the top spine**

````python
plt.plot(x, y)

ax = plt.gca()

ax.spines["top"].set_visible(False)
````

**Hide the right spine**

````python
plt.plot(x, y)

ax = plt.gca()

ax.spines["right"].set_visible(False)
````

**You can hide both:**

````python
plt.plot(x, y)

ax = plt.gca()

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.show()
````

This gives the graph a cleaner appearance.

## 13. Moving a Spine
You can also change the position of a spine.

For example:

````python
plt.plot(x, y)

ax = plt.gca()

ax.spines["left"].set_position(("data", 0))

plt.show()
````

This moves the left spine to x = 0.

And:

````python
plt.plot(x, y)

ax = plt.gca()

ax.spines["bottom"].set_position(("data", 0))

plt.show()
````

moves the bottom spine to y = 0.
    
This is more useful when creating mathematical coordinate plots, so you don't need to use it frequently in ordinary data analysis.

For now, remember:\
spines = borders of the plotting area

## 14. Tight Layout
Sometimes labels or titles can get cut off, especially when using rotated labels.

Matplotlib provides:
```python
plt.tight_layout()
```
Example:

````python
plt.figure(figsize=(8, 5))

plt.plot(
    months,
    sales,
    marker="o"
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
````

plt.tight_layout() automatically adjusts spacing so that plot elements fit better.

This is very useful when creating real plots.

## 15. A More Professional Plot
Let's combine several things:

````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 140, 180, 220, 200]

plt.figure(figsize=(10, 5))

plt.plot(
    months,
    sales,
    marker="o",
    lw=2,
    label="Sales"
)

plt.title(
    "Monthly Sales",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Month", fontsize=12)
plt.ylabel("Sales", fontsize=12)

plt.grid(
    linestyle="--",
    alpha=0.5
)

plt.annotate(
    "Highest Sales",
    xy=("May", 220),
    xytext=("Apr", 220),
    arrowprops=dict(arrowstyle="->")
)

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.legend()

plt.tight_layout()

plt.show()
````

This is getting much closer to a real data-analysis visualization.
