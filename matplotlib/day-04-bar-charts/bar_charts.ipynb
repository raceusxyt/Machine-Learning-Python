## 1. Basic Bar Chart — plt.bar()
Suppose we have:

````python
import matplotlib.pyplot as plt

categories = ["Food", "Travel", "Shopping", "Bills"]
amounts = [500, 300, 700, 400]

plt.bar(categories, amounts)

plt.show()
````

What does plt.bar() do?
```python
plt.bar(x, height)
```
Here:
- categories → x-axis categories
- amounts → height of each bar

So conceptually:
* Food      → 500
* Travel    → 300
* Shopping  → 700
* Bills     → 400

The tallest bar will be Shopping because its value is 700.

**Important difference from plt.plot()**

With a line plot:
```python
plt.plot(categories, amounts)
```
we're usually interested in trends.

With a bar chart:
```python
plt.bar(categories, amounts)
```
we're mainly interested in comparing categories.

## 2. Customizing Bar Charts
Just like line plots, we can customize bars.

**Bar color**

````python
plt.bar(subjects, marks, color="red")

plt.show()
````

**Bar width**

````python
plt.bar(subjects, marks, width=0.5)

plt.show()
````

The default width is approximately 0.8.

**Add labels and title**

````python
plt.bar(subjects, marks, color="green", width=0.6)

plt.title("Subject-wise Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.show()
````

So the general structure is:
```python
plt.bar(
    categories,
    values,
    color="...",
    width=...
)

plt.title(...)
plt.xlabel(...)
plt.ylabel(...)
plt.show()
```

## 3. Horizontal Bar Charts — plt.barh()
Sometimes category names are long, so a horizontal bar chart is easier to read.

Instead of:
```python
plt.bar(categories, values)
```
we use:
```python
plt.barh(categories, values)
```
Example:

````python
import matplotlib.pyplot as plt

categories = ["Food", "Transportation", "Entertainment", "Electricity"]
expenses = [500, 300, 200, 400]

plt.barh(categories, expenses)

plt.title("Monthly Expenses")
plt.xlabel("Amount")
plt.ylabel("Category")

plt.show()
````

The key difference:
```python
plt.bar()      # vertical
plt.barh()     # horizontal
```

## 4. Adding Values on Top of Bars
This is very useful in real data analysis.

Suppose:
```python
categories = ["Food", "Travel", "Shopping"]
amounts = [500, 300, 700]
```
We can display the actual values above each bar.

````python
import matplotlib.pyplot as plt

categories = ["Food", "Travel", "Shopping"]
amounts = [500, 300, 700]

bars = plt.bar(categories, amounts)

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        bar.get_height(),
        ha="center",
        va="bottom"
    )

plt.show()
````

**What's happening?**

plt.bar() returns the individual bar objects:
```python
bars = plt.bar(categories, amounts)
```
Then:
```python
for bar in bars:
```
lets us work with each bar.

Important properties:
```python
bar.get_height()
```
→ gets the height/value of the bar.
```python
bar.get_x()
```
→ gets the left x-position.
```python
bar.get_width()
```
→ gets the width of the bar.

So:
```python
bar.get_x() + bar.get_width() / 2
```
finds the center of the bar.

And:
```python
bar.get_height()
```
gives us the position where we want to place the value.

The two uses of get_height() are:
```python
bar.get_height()
```
First use
As the y-coordinate:
```python
bar.get_height(),
```
This tells Matplotlib: "Put the text at the top of the bar."

Second use
As the text itself:
```python
bar.get_height(),
```
This tells Matplotlib: "Display the bar's value."

So for a bar with height 500:
```python
plt.text(
	x_position,
	500,       # where to place the text
	500        # what text to display
)

Conceptually:
             500  ← text = 500
              ↑
              │
              █
              █
              █
              █
              █
              └── height = 500

A clearer version
We can avoid calling it twice by storing the value:

````python
import matplotlib.pyplot as plt

categories = ["Food", "Travel", "Shopping"]
amounts = [500, 300, 700]

bars = plt.bar(categories, amounts)

for bar in bars:
    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        height,
        ha="center",
        va="bottom"
    )

plt.show()
````

Now it's easier to understand:
```python
height = bar.get_height()
```
Then:
```python
height       # where to put the text
height       # what to write
```
So get_height() isn't doing two different calculations. We're simply using the same bar height for two purposes: position and displayed value.

## 5. Grouped Bar Charts
This is an important one for data analysis.

Suppose we want to compare sales of two products across months:
```python
months = ["Jan", "Feb", "Mar"]

product_a = [100, 120, 150]
product_b = [80, 140, 130]
```
We don't want the bars to overlap. We place them side-by-side.

````python
import numpy as np
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar"]

product_a = [100, 120, 150]
product_b = [80, 140, 130]

x = np.arange(len(months))
width = 0.35

plt.bar(x - width/2, product_a, width, label="Product A")
plt.bar(x + width/2, product_b, width, label="Product B")

plt.xticks(x, months)

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Product Sales")

plt.legend()

plt.show()
````

The important idea
```python
x = np.arange(len(months))
```
Since there are 3 months:
```python
x = [0, 1, 2]
```
Then:
```python
x - width/2
```
moves Product A slightly left.
And:
```python
x + width/2
```
moves Product B slightly right.

So we get:
```text
 A  B         A  B         A  B
 █  █         █  █         █  █
 █  █         █  █         █  █
  Jan          Feb          Mar
```
This is called a grouped bar chart.

## 7. Stacked Bar Charts
A stacked bar chart combines multiple values into a single bar, showing both the individual contributions and the total.

Suppose we track two types of expenses:

````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]

food = [200, 250, 180, 300]
travel = [100, 150, 120, 200]

plt.bar(months, food, label="Food")
plt.bar(months, travel, bottom=food, label="Travel")

plt.title("Monthly Expenses by Category")
plt.xlabel("Month")
plt.ylabel("Amount")

plt.legend()

plt.show()
````

The key concept: bottom=food
- The first plt.bar() draws the Food bars from zero.
- The second plt.bar() draws the Travel bars on top of the Food bars.
- bottom=food tells Matplotlib where each Travel bar should begin.

For January:
- Food = 200
- Travel = 100
- Total height = 300
