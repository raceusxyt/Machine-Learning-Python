# Day 5 — Scatter Plots 📊
Today we'll learn how to visualize relationships between numerical variables. Scatter plots are especially useful in data analysis and machine learning because they help us spot patterns, clusters, and relationships.

## 1. What is a scatter plot?
A scatter plot displays individual data points using two numerical variables:
- X-axis: the first variable.
- Y-axis: the second variable.
- Each dot: one observation.

For example, suppose we record how many hours students study and the marks they receive.

## 2. Creating a scatter plot with plt.scatter()
Let's start with code.

````python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6]
marks = [35, 40, 50, 55, 70, 80]

plt.scatter(hours, marks)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
````

**How does it work?**
```python
plt.scatter(x, y)
```
For our example:
- hours supplies the x-coordinates.
- marks supplies the y-coordinates.
- The first point is (1, 35).
- The second point is (2, 40).
- The last point is (6, 80).

Unlike plt.plot(), plt.scatter() displays individual points by default, without connecting them with lines.

## 3. Customizing scatter plots
Just like bar and line charts, we can customize scatter plots.

**A. Change point color and size**

````python
plt.scatter(
    temperature,
    ice_cream_sales,
    color="red",
    s=100
)

plt.xlabel("Temperature")
plt.ylabel("Ice Cream Sales")

plt.show()
````

- color="red" changes the point color.
- s=100 sets the marker area (in points squared). Larger values create larger dots.

**B. Change marker shape**

````python
plt.scatter(
    temperature,
    ice_cream_sales,
    color="green",
    marker="^",
    s=100
)

plt.xlabel("Temperature")
plt.ylabel("Ice Cream Sales")

plt.show()
````

Common marker shapes:
Marker	Shape
- "o"	Circle (default)
- "s"	Square
- "^"	Upward triangle
- "x"	Cross
- "*"	Star

**C. Add transparency**

````python
plt.scatter(
    temperature,
    ice_cream_sales,
    color="blue",
    marker="s",
    alpha=0.5,
    s=120
)

plt.xlabel("Temperature")
plt.ylabel("Ice Cream Sales")

plt.show()
````

alpha controls transparency:
- alpha=1 → fully opaque
- alpha=0.5 → semi-transparent
- alpha=0 → fully transparent

Transparency is especially useful when many points overlap.

## 5. Adding colors based on categories
Sometimes we want different colors for different groups—for example, students who passed versus students who failed.
```python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6]
marks = [25, 35, 45, 55, 75, 90]

colors = ["red", "red", "red", "green", "green", "green"]

plt.scatter(hours, marks, c=colors, s=100)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

Here, c=colors assigns a color to each point:
- Red → first three students
- Green → last three students

We can also add a legend to explain the colors, which we'll practice later.

## 6. A useful shortcut: color mapping with numerical values
Instead of manually assigning colors, we can use a numerical variable to control point colors.

````python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6]
marks = [25, 35, 45, 55, 75, 90]

plt.scatter(hours, marks, c=marks, cmap="viridis", s=100)

plt.colorbar(label="Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")

plt.show()
````

Here:
- c=marks maps each point's color to its marks value.
- cmap="viridis" selects the color palette.
- plt.colorbar() shows what the colors mean.

This is useful when visualizing three variables at once: X position, Y position, and color.

## 🧠 Final Day 5 Challenge
Let's combine what you've learned into one practical scatter plot.

Imagine you're analyzing students' study habits:
```python
hours = [1, 2, 3, 4, 5, 6, 7, 8]
marks = [25, 35, 40, 50, 60, 72, 85, 92]
```

Create a scatter plot with these requirements:
1. X-axis: hours; Y-axis: marks.
2. Marker color based on marks, using cmap="viridis".
3. Marker size: 100.
4. Add a color bar labeled "Marks".
5. Title: "Study Hours vs Marks".
6. X-axis label: "Study Hours".
7. Y-axis label: "Marks".
8. Add a grid with alpha=0.3.
9. Display the plot using plt.show().

````python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6, 7, 8]
marks = [25, 35, 40, 50, 60, 72, 85, 92]

plt.scatter(hours, marks, c=marks, cmap="viridis", s=100)

plt.colorbar(label="Marks")
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.grid(alpha=0.3)

plt.show()
````

One important takeaway: scatter plots help you explore relationships between numerical variables. They're especially useful for investigating correlations and spotting patterns before building a machine learning model.
