# Day 06: Histograms
Histograms are especially useful in data analysis because they help us understand how values are distributed.

## 1. What is a histogram?
A histogram shows the distribution of numerical data by grouping values into intervals called bins and displaying how many values fall into each interval.

<div>
	<img src="../Images/Histogram.png" style="border-radius: 30px; max-width: 80%;">
</div>

For example, suppose these are the marks of 15 students:
```python
marks = [12, 18, 22, 25, 28, 32, 35, 38,
         42, 45, 48, 52, 65, 72, 88]
```
A histogram helps us answer questions like:
- Which range contains the most students?
- Are most marks low, medium, or high?
- Are the marks concentrated in one range or spread out?

## 2. Your first histogram — plt.hist()

Run this code:

````python
import numpy as np
import matplotlib.pyplot as plt

marks = [12, 18, 22, 25, 28, 32, 35, 38,
         42, 45, 48, 52, 65, 72, 88]

plt.hist(marks)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
````

The important function is:
```python
plt.hist(marks)
```
Unlike a bar chart, where you usually supply categories and their values, a histogram receives the raw numerical observations and groups them into intervals automatically.

## 3. Understanding bins
The bins parameter controls how many intervals the data is divided into.
```python
plt.hist(marks, bins=5)
```
This groups the marks into five intervals. Each bar's height tells you how many marks fall inside that interval.

For example, if one interval contains 20, 25, 28, 29, its frequency is 4.

Try these separately to compare the results:

````python
plt.hist(marks, bins=5)

plt.show()
````

````python
plt.hist(marks, bins=10)

plt.show()
````

More bins give you a more detailed view of the distribution, while fewer bins provide a broader overview.

## 5. Customize a histogram
You can control the appearance of a histogram using parameters such as color, edgecolor, and bins.

````python
import matplotlib.pyplot as plt

marks = [12, 18, 22, 25, 28, 32, 35, 38,
         42, 45, 48, 52, 65, 72, 88]

plt.hist(
    marks,
    bins=5,
    color="red",
    edgecolor="black"
)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.grid(axis="y", alpha=0.3)

plt.show()
````

Here's what the new parameters do:
| Parameter | Purpose |
| :--- | :--- |
| bins=5 | Divides the data into five intervals |
| color="red" | Sets the bar fill color to red |
| edgecolor="black" | Adds black outlines to the bars |
| plt.grid(axis="y") |Adds horizontal grid lines to help compare frequencies |

Notice that axis="y" adds grid lines only in the vertical direction of the chart's values, making bar heights easier to compare.

## 6. Control the bin intervals
You don't have to let Matplotlib choose the intervals automatically. You can specify the exact bin edges yourself.

````python
plt.hist(
    marks,
    bins=[0, 20, 40, 60, 80, 100],
    color="orange",
    edgecolor="black"
)

plt.title("Student Marks by Range")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.grid(axis="y", alpha=0.5)

plt.show()
````

Here, the bins represent these ranges:
- 0–20
- 20–40
- 40–60
- 60–80
- 80–100

Typically, each bin includes its left edge but excludes its right edge, except the **final bin**, which includes the final edge. So a mark of 40 belongs to the 40–60 bin, not the 20–40 bin.

This is useful when you want meaningful ranges, such as grade bands or spending categories.

## 7. Understand density=True
So far, our y-axis has shown frequency—how many observations are in each bin.

What if you want to visualize the distribution on a probability-density scale instead?

````python
plt.hist(
    marks,
    bins=5,
    density=True,
    edgecolor="black"
)

plt.title("Density Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Density")

plt.show()
````

Setting density=True normalizes the histogram so that the total area of all bars equals 1. The bar heights are density values, not the number of students or probabilities directly.

For now, remember:
- Default density=False: shows frequencies.
- density=True: shows probability density, useful for comparing distributions on a normalized scale.

## 10. Final concept: Compare frequency and density

````python
plt.hist(expenses, bins=5, edgecolor="black")
plt.title("Frequency Histogram")
plt.ylabel("Frequency")

plt.show()
````

This shows the number of expenses in each bin.

Compare it with:
