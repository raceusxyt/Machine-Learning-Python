# Day 08 — Matplotlib Subplots
Today, you'll learn how to display multiple charts in a single figure. This is useful in data analysis when you want to compare sales, expenses, marks, or other datasets side by side.

## 1. What are subplots?
A subplot is an individual chart inside a larger figure.

	<img src="../Images/Subplots.png" style="border-radius: 30px; max-width: 80%;">

For example, instead of creating separate figures for sales and expenses, we can place both charts in the same figure.

## 2. Your first subplot
The main function is plt.subplots().

````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 130, 180, 220]
expenses = [80, 90, 100, 120, 140]

fig, ax = plt.subplots(1, 2, figsize=(10, 4))

ax[0].plot(months, sales, marker="o")
ax[0].set_title("Monthly Sales")
ax[0].set_xlabel("Month")
ax[0].set_ylabel("Sales")

ax[1].plot(months, expenses, marker="s", color="orange")
ax[1].set_title("Monthly Expenses")
ax[1].set_xlabel("Month")
ax[1].set_ylabel("Expenses")

plt.tight_layout()
plt.show()
````

**Understand the important parts**
- plt.subplots(1, 2) creates 1 row and 2 columns of charts.
- fig represents the entire figure.
- ax contains the individual subplot axes.
- ax[0] accesses the first chart.
- ax[1] accesses the second chart.
- ax[0].plot() draws on the first chart.
- ax[1].plot() draws on the second chart.
- set_title(), set_xlabel() and set_ylabel() customize each subplot.
- plt.tight_layout() helps prevent overlapping labels and titles.

Remember: When using ax, use methods such as ax[0].set_title() rather than plt.title() to customize an individual subplot.

## 3. Subplots with two rows and two columns
You can create four charts using plt.subplots(2, 2).

````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 150, 130, 180]
expenses = [70, 90, 85, 110]
marks = [60, 75, 68, 90]

fig, ax = plt.subplots(2, 2, figsize=(10, 7))

# Top-left: line plot
ax[0, 0].plot(months, sales, marker="o")
ax[0, 0].set_title("Sales")

# Top-right: bar chart
ax[0, 1].bar(months, expenses, color="orange")
ax[0, 1].set_title("Expenses")

# Bottom-left: scatter plot
ax[1, 0].scatter(sales, marks)
ax[1, 0].set_title("Sales vs Marks")
ax[1, 0].set_xlabel("Sales")
ax[1, 0].set_ylabel("Marks")

# Bottom-right: histogram
ax[1, 1].hist(marks, bins=4, edgecolor="black")
ax[1, 1].set_title("Marks Distribution")

plt.tight_layout()
plt.show()
````

For a 2 × 2 layout, access the charts using two indices:
| Subplot | Index |
| :--- | :--- |
| Top-left | ax[0, 0] |
| Top-right | ax[0, 1] |
| Bottom-left | ax[1, 0] |
| Bottom-right | ax[1, 1] |

## 4. Sharing axes
When comparing charts with similar scales, you can share their axes.
```python
fig, ax = plt.subplots(1, 2, sharey=True, figsize=(10, 4))
```
- sharey=True: both charts share the same y-axis scale.
- sharex=True: both charts share the same x-axis scale.

This makes comparisons fairer because the charts use consistent scales.

## 6. Final coding challenge
Create a figure with four subplots using the data below:
```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [120, 150, 140, 180, 220]
expenses = [80, 90, 100, 110, 130]
profit = [40, 60, 40, 70, 90]
```
Requirements:
1. Use plt.subplots(2, 2, figsize=(10, 7)).
2. Top-left: line chart of sales.
3. Top-right: bar chart of expenses.
4. Bottom-left: line chart of profit.
5. Bottom-right: histogram of sales.
6. Add a title to each subplot and call plt.tight_layout() before plt.show().

````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [120, 150, 140, 180, 220]
expenses = [80, 90, 100, 110, 130]
profit = [40, 60, 40, 70, 90]

fig, ax = plt.subplots(2, 2, figsize=(10, 7))

ax[0, 0].plot(months, sales, marker="o", color="blue")
ax[0, 0].set_title("Sales")
ax[0, 0].set_xlabel("Month")
ax[0, 0].set_ylabel("Sales")

ax[0, 1].bar(months, expenses, color="red")
ax[0, 1].set_title("Expenses")
ax[0, 1].set_xlabel("Month")
ax[0, 1].set_ylabel("Expenses")

ax[1, 0].plot(months, profit, marker="s", color="red")
ax[1, 0].set_title("Profit")
ax[1, 0].set_xlabel("Month")
ax[1, 0].set_ylabel("Profit")

ax[1, 1].hist(sales, color="blue", edgecolor="black")
ax[1, 1].set_title("Sales Distribution")
ax[1, 1].set_xlabel("Sales")
ax[1, 1].set_ylabel("Frequency")

plt.tight_layout()

plt.show()
````
