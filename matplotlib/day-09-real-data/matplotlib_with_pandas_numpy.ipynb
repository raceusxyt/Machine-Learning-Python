## 2. NumPy + Matplotlib
NumPy is useful for creating and transforming numerical data before visualization.

````python
import numpy as np
import matplotlib.pyplot as plt

months = np.array(["Jan", "Feb", "Mar", "Apr", "May", "Jun"])
sales = np.array([120, 150, 140, 180, 200, 240])
expenses = np.array([80, 90, 100, 110, 120, 130])

profit = sales - expenses

plt.figure(figsize=(9, 5))

plt.plot(
    months,
    sales,
    marker="o",
    label="Sales"
)
plt.plot(
    months,
    expenses,
    marker="s",
    label="Expenses"
)
plt.plot(
    months,
    profit,
    marker="^",
    label="profit"
)

plt.title("Monthly Business Performance")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.grid(axis="y", alpha=0.3)

plt.legend()

plt.show()
````

Understand the important part
```python
profit = sales - expenses
```
NumPy subtracts corresponding elements automatically.

For example:
- January: 120 - 80 = 40
- February: 150 - 90 = 60

This is called element-wise arithmetic. You don't need a loop.

## 3. Pandas DataFrame + Matplotlib
Now let's store the data in a DataFrame.

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "sales": [120, 150, 140, 180, 200, 240],
    "expenses": [80, 90, 100, 110, 120, 130]
})

df["profit"] = df["sales"] - df["expenses"]

df
````

You should get a DataFrame with four columns: month, sales, expenses, and profit.

**Plot directly from the DataFrame**
Pandas has built-in plotting methods that use Matplotlib behind the scenes.

````python
df.plot(
    x="month",
    y=["sales", "expenses", "profit"],
    kind="line",
    marker="o",
    figsize=(9, 5)
)

plt.title("Monthly Business Performance")
plt.ylabel("Amount")
plt.grid(axis="y", alpha=0.3)

plt.show()
````

Here, kind="line" selects a line chart. Other useful values include "bar", "scatter", and "hist".

## 4. Category-wise expense analysis
Let's use a dataset similar to the expense-analysis project you practiced earlier.

````python
expenses_df = pd.DataFrame({
    "category": [
        "Food", "Travel", "Food", "Shopping",
        "Travel", "Bills", "Food", "Shopping"
    ],
    "amount": [250, 100, 300, 500, 150, 800, 200, 350]
})

category_totals = expenses_df.groupby("category")["amount"].sum()

print(category_totals)
````

Visualize the totals with a bar chart:

````python
category_totals.plot(
    kind="bar",
    color="skyblue",
    edgecolor="black",
    figsize=(8, 5)
)

plt.title("Total Expenses by Category")
plt.xlabel("Category")
plt.ylabel("Total Amount")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.show()
````

What is happening?
1. groupby("category") groups the rows by expense category.
2. ["amount"].sum() adds the amounts in each category.
3. .plot(kind="bar") visualizes the resulting totals.

This is a common workflow in exploratory data analysis (EDA).

## 5. NumPy statistics + visualization
NumPy can calculate statistics that help you understand a dataset before plotting it.

````python
import numpy as np
import matplotlib.pyplot as plt

daily_expenses = np.array([
    120, 150, 180, 200, 220,
    250, 280, 300, 350, 500
])

print("Mean:", np.mean(daily_expenses))
print("Median:", np.median(daily_expenses))
print("Maximum:", np.max(daily_expenses))
print("Minimum:", np.min(daily_expenses))

plt.hist(
    daily_expenses,
    bins=5,
    edgecolor="black",
    color="lightgreen",
)
plt.axvline(
    np.mean(daily_expenses),
    color="red",
    linestyle="--",
    label="Mean"
)

plt.title("Daily Expense Distribution")
plt.xlabel("Expense Amount")
plt.ylabel("Frequency")

plt.legend()

plt.show()
````

plt.axvline() draws a vertical reference line. Here, it marks the mean expense so you can compare the distribution with its average.

For, horizontal reference line: plt.axhline()

Build a small business dashboard using the following data:
```python
df = pd.DataFrame({

    "sales": [200, 250, 220, 300, 350, 400],

})
```

Your tasks:
1. Create a profit column using Pandas.
2. Print the mean sales using NumPy.
3. Create a figure with two subplots in one row.
4. On the first subplot, plot sales and expenses as two lines.
5. On the second subplot, create a bar chart of monthly profit.
6. Add titles, axis labels, and a legend where appropriate.
7. Use plt.tight_layout() and plt.show().

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "sales": [200, 250, 220, 300, 350, 400],
    "expenses": [120, 140, 150, 180, 200, 220]
})

df["profit"] = df["sales"] - df["expenses"]

print("Mean:", df["sales"].mean())

fig, ax = plt.subplots(1, 2, figsize=(10, 5))

ax[0].plot(
    df["month"],
    df["sales"],
    marker="o",
    color="blue",
    linestyle="-",
    lw=2,
    label="Sales"
)

ax[0].plot(
    df["month"],
    df["expenses"],
    marker="s",
    color="red",
    linestyle="--",
    lw=2,
    label="Expenses"
)

ax[0].set_title("Sales vs Expenses")
ax[0].set_xlabel("Months")
ax[0].set_ylabel("Sales and Expenses")

ax[0].legend()

ax[1].bar(
    df["month"],
    df["profit"],
    color="lightgreen",
    edgecolor="black"
)

ax[1].set_title("Monthly Profit")
ax[1].set_xlabel("Months")
ax[1].set_ylabel("Profit")

plt.tight_layout()

plt.show()
````
