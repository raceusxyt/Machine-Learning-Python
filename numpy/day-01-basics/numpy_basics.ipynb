# 🐍 Day 1 — Introduction to NumPy
## What is NumPy?

**NumPy (Numerical Python)** is a Python library designed for numerical and scientific computing.

Its most important feature is the **NumPy array**, called an ndarray.

Python list:
```python
numbers = [10, 20, 30, 40, 50]

NumPy array:

import numpy as np

numbers = np.array([10, 20, 30, 40, 50])
```
The NumPy version allows us to perform mathematical operations efficiently on the entire array.

For example:
```python
numbers * 2

Output:

[ 20  40  60  80 100]
```
Notice that we didn't need a loop.

## 1️⃣ Import NumPy

The standard way is:
```python
import numpy as np
```
**np** is simply an alias for NumPy.

So instead of writing:
```python
numpy.array(...)

we normally write:

np.array(...)
```

### 1️⃣ np.zeros()

Creates an array filled with 0.

````python
import numpy as np

arr_1d = np.zeros(5)
print("1D Array:", arr_1d)

print()

arr_2d = np.zeros((2, 3))
print("2D Array:\n", arr_2d)
````

### 2️⃣ np.ones()

Exactly the same idea, but filled with 1.

````python
arr_1d = np.ones(5)
print(f"1D Array: {arr_1d}")

print()

arr_2d = np.ones((2, 3))
print(f"2D Array:\n {arr_2d}")
````

### 3️⃣ np.arange()

This one is particularly useful.

It works similarly to Python's range().

````python
arr = np.arange(1, 6)

print(arr)
````

**Important rule**
```python
np.arange(start, stop)
```
The stop value is NOT included.

So:
```python
np.arange(1, 6)
```
means:
```text
1, 2, 3, 4, 5

not 6.

You can also specify a step:
```
```python
np.arange(0, 11, 2)
```
Output:
```python
[ 0  2  4  6  8 10]
```
Meaning:

* start = 0
* stop  = 11
* step  = 2

````python
arr = np.arange(0, 11, 2)

print(arr)
````

### 4️⃣ np.linspace()

This one is slightly different.

linspace() creates a specified number of evenly spaced values between two numbers.

````python
arr = np.linspace(0, 10, 5)

print(arr)
````

Here:
```text
start = 0
stop  = 10
number of values = 5
```
Unlike arange(), the stop value is included by default.

### 🧠 arange() vs linspace()

This distinction is important.

arange()

You specify the step:
```python
np.arange(0, 10, 2)

→

0 2 4 6 8
```
linspace()

You specify the number of values:
```python
np.linspace(0, 10, 6)

→

0 2 4 6 8 10

So remember:

arange → "How big should the steps be?"
linspace → "How many values do I want?"
```

````python
arr = np.linspace(0, 10, 6)

print(arr)
````

## 🧭 Understanding axis

axis tells NumPy which direction to perform an operation.

Consider this 2D array:

````python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
````

axis=0, Thinks down the rows:

````python
np.sum(arr, axis=0)
````

Calculates:
```python
10 + 40 + 70 = 120
20 + 50 + 80 = 150
30 + 60 + 90 = 180
```
Result:
```python
[120 150 180]

So axis=0 gives you column-wise results.
```

axis=1, Thinks across the columns:

````python
np.sum(arr, axis=1)
````

Calculates:
```python
10 + 20 + 30 = 60
40 + 50 + 60 = 150
70 + 80 + 90 = 240
```
Result:
```python
[ 60 150 240]

So axis=1 gives you row-wise results.
```

### 🧠 Easy way to remember

For a 2D array:

* axis=0 → ↓ down → columns
* axis=1 → → across → rows

# 🐍 Next: Multiple Conditions

NumPy also lets us combine conditions.

Suppose:

````python
arr = np.array([10, 20, 30, 40, 50])
````

We want values greater than 20 AND less than 50.

We can write:

and: |, for OR

rather than Python's:
```text
and
or
```

```python

means:

greater than 20 AND less than 50.
```

# 🐍 Next: where() — Finding/Changing Values

NumPy provides another useful function: np.where()

It lets us find positions based on a condition or create values based on a condition.

````python
arr = np.array([10, 20, 30, 40, 50])

np.where(arr > 30)
````

Because:
```text
Index:   0   1   2   3   4
Value:  10  20  30  40  50
                     ↑   ↑
                     3   4

So np.where(arr > 30) tells us:

"At which indexes is the condition true?"
```

### 🔄 np.where() can also replace values

This form is extremely useful:
```python
np.where(condition, value_if_true, value_if_false)
```

````python
arr =np.array([10, 20, 30, 40, 50])

result = np.where(arr > 30, 100, 0)

print(result)
````

Meaning:
```python
10 > 30 → False → 0
20 > 30 → False → 0
30 > 30 → False → 0
40 > 30 → True  → 100
50 > 30 → True  → 100
```
The original arr isn't changed.

### 🧠 Why this matters

You'll eventually encounter patterns like:
```python
np.where(expenses > 1000, "High", "Normal")
```
which can classify an entire array at once.

That's very similar to the kind of logic you already used in your Pandas expense analysis.

# 📐 Next: Reshaping NumPy Arrays

This is an important topic because data often needs to be converted between different shapes.

````python
arr = np.array([1, 2, 3, 4, 5, 6])
````

````python
arr.shape
````

We can turn it into a 2 × 3 array:

````python
arr2 = arr.reshape(2, 3)

print(arr2)
````

````python
arr2.shape
````

### 🧠 Important rule

The total number of elements must remain the same.

Our original array has:

6 elements

So these are valid:
```python
arr.reshape(2, 3)   # 2 × 3 = 6
arr.reshape(3, 2)   # 3 × 2 = 6
arr.reshape(1, 6)   # 1 × 6 = 6
arr.reshape(6, 1)   # 6 × 1 = 6
```
But this is invalid:
```python
arr.reshape(2, 4)
```
because: 2 × 4 = 8

and we only have 6 elements.

### 📐 reshape() — One Useful Trick

NumPy can automatically calculate one dimension using -1.

For example:

````python
arr = np.array([10, 20, 30, 40, 50, 60, 70, 80])

arr.reshape(2, -1)
````

**NumPy sees:**
```text
8 elements
2 rows

and calculates:

8 ÷ 2 = 4 columns

So the result is:
```
```text
[[10 20 30 40]
 [50 60 70 80]]
```
Similarly:
```python
arr.reshape(-1, 2)
```
produces:
```python
[[10 20]
 [30 40]
 [50 60]
 [70 80]]
```
### 🧠 Rule

You can use only one -1 in a reshape.
```text
reshape(2, -1)     # ✅
reshape(-1, 2)     # ✅
reshape(2, -1, 2)  # ❌ generally ambiguous
```

# 🚀 Next: Broadcasting

This is one of the most important NumPy concepts.

You already used a simple form of broadcasting:

````python
arr = np.array([10, 20, 30])

arr + 5
````

You can think of NumPy as treating the 5 as if it were applied to every element:
```python
[10 20 30]
 +5 +5 +5
------------
[15 25 35]
```
But broadcasting becomes more interesting with different-shaped arrays.

For example:

````python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

arr + np.array([1, 2, 3])
````

The 1D array:
```text
[1 2 3]
```
**is applied to each row.**

This is broadcasting.

Variance is related to standard deviation:

So:
```python
np.var(arr)
```
is equivalent to approximately:
