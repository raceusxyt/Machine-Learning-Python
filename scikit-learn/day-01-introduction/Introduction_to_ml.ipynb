## 1. What is Machine Learning?
Machine learning (ML) is a branch of artificial intelligence that enables computers to learn patterns from data and use those patterns to make predictions or decisions, rather than requiring us to manually write every rule.

Let's understand this with an example.

Imagine you want to build a program that predicts the price of a house.

A house's price may depend on several factors:
- Area of the house in square feet
- Number of bedrooms
- Location
- Age of the property

Instead of manually writing rules for every possible house, we can give a machine learning algorithm examples of houses with their known prices.

The algorithm learns patterns from those examples and uses the learned patterns to estimate the price of a new house.

For example:
| House area | Known price |
| :--- | :--- |
| 800 sq. ft. |₹20 lakh |
| 1,000 sq. ft. | ₹25 lakh |
| 1,200 sq. ft. | ₹30 lakh |
| 1,500 sq. ft. | ₹37 lakh |

After learning from this data, a model might estimate the price of a 1,100 sq. ft. house at approximately ₹27.5 lakh, assuming the other factors are comparable.

This is an illustrative example, not a real-estate valuation.

**The key idea:** data → learning patterns → model → prediction.

## 5. Important machine learning terminology
These terms will appear throughout our lessons, so let's understand them now.
| Term | Meaning | Example |
| :--- | :--- | :--- |
| Dataset | Collection of examples | Records of houses |
| Sample / instance | One example | One house |
| Feature | Input variable | Area, bedrooms |
| Target / label | Value to predict | House price |
| Algorithm | Learning procedure | Linear regression algorithm |
| Model | Learned relationship | Fitted regression model |
| Training | Learning from data | Fitting the model |
| Prediction | Output for new input | Estimated house price |

**Features and targets in a dataset**
Suppose we have the following data:
| Area (sq. ft.) | Bedrooms | Price (₹ lakh) |
| :--- | :--- | :--- |
| 800 | 2 | 20 |
| 1,000 | 2 | 25 |
| 1,200 | 3 | 30 |
| 1,500 | 3 | 37 |

Here:
- **Features**, **X**: area and bedrooms.
- **Target**, **y**: price.
- **Samples**: four houses.
- **Number of features**: two.
In Python, we often use `X` for the input features and `y` for the target.
```python
X = [
    [800, 2],
    [1000, 2],
    [1200, 3],
    [1500, 3]
]

y = [20, 25, 30, 37]
```
Notice that `X` is a two-dimensional structure: each row is a house, and each column represents a feature. `y` contains one target value for each house.

This distinction will become very important when we use Scikit-learn.

## 7. Your first machine learning program
Now let's use Scikit-learn to train a model and make a prediction.

We'll begin with **Linear Regression**, one of the simplest supervised learning algorithms. We'll study its mathematics properly in a later lesson; today, our goal is to understand the basic workflow.

### Explanation
Imagine that house prices follow this simplified relationship:
$$
\text{Price} = 0.025 \times \text{Area}
$$

For a house of 800 square feet, the price would be $$0.025 \times 800 = 20 \ lakh.$$

We'll give the model four examples and ask it to estimate the price of a 1,100-square-foot house.

### Code
**Import the library and prepare the data.**

````python
from sklearn.linear_model import LinearRegression

# Training data
X = [[800], [1000], [1200], [1500]]
y = [20, 25, 30, 37.5]

# Create the model
model = LinearRegression()

# Train the model
model.fit(X, y)

# Predict the price of a new house
prediction = model.predict([[1100]])

print("Predicted price:", prediction[0], "lakh")
````

| `LinearRegression()` | Creates a model object. |
| `model.fit(X, y)` | Learns the relationship between area and price. |
| `model.predict([[1100]])` |Predicts the price for a new house. |
| `prediction[0]` | Extracts the first predicted value, which is in this case `27.5` lakh |

**Why do we write `[[1100]]` rather than `[1100]`?**\
Scikit-learn expects the input features for prediction in a two-dimensional structure: rows represent samples and columns represent features. Here, we have one house and one feature.

**Important:**\
the model isn't thinking like a human. It estimates mathematical parameters from the training examples. This simple example uses perfectly linear data; real house prices are much more complicated.

````python
import pandas as pd

df = pd.DataFrame({
    "study_hours": [5.0, 5.5, 6.0, 6.5, 7.0, 7.5, 8.0, 8.5, 9.0, 9.5],
    "marks": [50, 54, 62, 68, 70, 75, 84, 88, 89, 95]
})

df
````

## 6. Train a linear regression model to predict the marks for 7.3 study-hours.

````python
from sklearn.linear_model import LinearRegression

X = df[["study_hours"]]
y = df["marks"]

model = LinearRegression()
model.fit(X, y)

prediction = model.predict(
    pd.DataFrame({"study_hours": [7.3]})
)

print("Predicted marks:", prediction[0].round(2))
````

Notice that `df[["study_hours"]]` uses double square brackets to produce a two-dimensional DataFrame, which is the format Scikit-learn expects for features.

<span style="color: #FF3333; font-weight: bold; font-size: 30px">Important:</span><br>
If you do:
```python
prediction = model.predict([[7.3]])
```
then it will give this warning:
```text
UserWarning: X does not have valid feature names, but LinearRegression was fitted with feature names
  warnings.warn(
```
<span style="color: #DA70D6; font-weight: bold; font-size: 20px">Why does this warning occur?</span><br>
You trained your model using:
```python
X = df[["study_hours"]]
y = df["marks"]

model = LinearRegression()
model.fit(X, y)
```
Here, `X` is a Pandas DataFrame with a named feature: `study_hours`.

But when you predicted using:
```python
prediction = model.predict([[7.3]])
```
The input `[[7.3]]` is a regular Python list, which doesn't contain feature names. Scikit-learn therefore warns you that the prediction input doesn't have the same feature names as the training data.

<span style="color: #33FF33; font-weight: bold; font-size: 20px">Solution 1 — Recommended</span><br>
Use a DataFrame with the same column name when predicting.
```python
prediction = model.predict(
    pd.DataFrame({"study_hours": [8.0]})
)

print("Predicted marks:", prediction[0])
```
which we used in our case. This is the cleanest solution because the feature name matches the training data.

<span style="color: #33FF33; font-weight: bold; font-size: 20px">Solution 2 — Use NumPy-style input consistently</span><br>
You can also train the model using a NumPy array instead of a DataFrame.
```python
X = df[["study_hours"]].to_numpy()
y = df["marks"]

model = LinearRegression()
model.fit(X, y)

prediction = model.predict([[7.3]])

print("Predicted marks:", prediction[0])
```
Now both training and prediction use inputs without feature names, so the warning should disappear.

**Remember:**
- `fit()` learns from the training data.
- `predict()` expects input features in a compatible format.
- Matching feature names is especially useful when working with Pandas DataFrames.
