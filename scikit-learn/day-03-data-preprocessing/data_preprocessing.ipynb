## 1. Understanding data preprocessing
### What is data preprocessing?
**Data preprocessing** is the process of cleaning and transforming raw data into a suitable form for analysis and machine learning.

Consider this student dataset:
| Study hours | Attendance (%) | Department | Marks |
| :--- | :--- | :--- | :--- |
| 5 | 80 | CSE | 65 |
| 7 | 90 | ECE | 82 |
| 6 | Missing | CSE | 74 |
| 8 | 95 | Missing | 90 |
| 7 | 90 | ECE | 82 |

**What problems can you identify?**
- Attendance is missing for one student.
- Department is missing for another.
- The last row duplicates an earlier record.
- Department contains text rather than numerical values.

These are common data-quality problems. Later, we'll also examine scaling and outliers.

**Why is preprocessing important?**
- Many machine learning algorithms cannot handle missing values directly.
- Most Scikit-learn estimators require categorical text to be encoded appropriately.
- Features on very different scales can affect some algorithms, such as K-nearest neighbors and support vector machines.
- Duplicate or incorrectly recorded observations can distort analysis.
- Poor preprocessing can cause data leakage if information from the test set influences the training process.

<span style="color: #FF3333; font-weight: bold; font-size: 20px">Important:</span><br>
preprocessing does not mean blindly modifying every dataset. First inspect the data, understand the problem, and then choose an appropriate treatment.

### Let's create a practice dataset
We'll use Pandas to create a small dataset containing missing values, categorical data, and a duplicate row.

**Code**

**Let's understand the mean vs. median**
Suppose attendance percentages are:

`70, 80, 90, 95, 100`

The mean is:
$$\frac{70+80+90+95+100}{5}=87$$

The median is `90`, the middle value after sorting.

Now suppose one value is an extreme outlier:

`70, 80, 90, 95, 500`

The mean becomes `167`, while the median remains `90`. The median is much less affected by this extreme value.

Of course, attendance cannot realistically be 500%, so this example also illustrates why data validation matters.

**Scikit-learn's** `SimpleImputer`
For machine learning, Scikit-learn provides a reusable imputation transformer.

````python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy="median")

attendance_filled = imputer.fit_transform(
    df[["attendance"]]
)

print(attendance_filled)
````

Here, `fit_transform()` calculates the median from the supplied data and fills in the missing values.

**Critical rule for ML:** once you have a training/test split, fit the imputer only on training data. Then use that fitted imputer to transform the test data. Otherwise, the test set can leak information into training.

## 4. Encoding categorical variables
Machine learning algorithms generally need numerical feature representations. Our `department` column contains text categories such as `CSE`, `ECE`, and `ME`.

Encoding converts categories into a numerical representation.

### Method A — One-hot encoding
One-hot encoding creates a separate binary column for each category.

For example:
| Department | CSE | ECE | ME |
| :--- | :--- | :--- | :--- |
| CSE | 1 | 0 | 0 |
| ECE | 0 | 1 | 0 |
| ME | 0 | 0 | 1 |

This avoids incorrectly implying that one department is numerically greater than another.

**Code using Pandas**

````python
encoded_df = pd.get_dummies(
    df_cleaned,
    columns=["department"],
    dtype=int
)

encoded_df
````

### Method B — `OneHotEncoder` in Scikit-learn
For a reusable machine learning workflow:

````python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

encoded = encoder.fit_transform(
    df_cleaned[["department"]]
)

print(encoder.get_feature_names_out(["department"]))

print(encoded)
````

`handle_unknown="ignore"` allows the encoder to handle a category that wasn't observed when it was fitted.

This example assumes a recent Scikit-learn version. In older versions, the parameter was called `sparse` rather than `sparse_output`.

### Method C — Ordinal encoding
Ordinal encoding is appropriate when categories have a genuine order, such as:
- Low
- Medium
- High

These might be represented as `0`, `1`, and `2`.

However, assigning arbitrary numbers to nominal categories such as departments or colors can mislead algorithms into treating the numbers as meaningful quantities.

**Rule of thumb:**\
use one-hot encoding for nominal categories; use ordinal encoding when a meaningful ordering exists and is represented correctly.

## 5. Feature scaling: normalization vs. standardization
Imagine that your dataset has two features:
- Study hours: `1–10`
- Annual family income: `100,000–2,000,000`

The numerical ranges are very different. Some algorithms are sensitive to feature scales, so we may need to scale the features.

Scaling is particularly useful for K-nearest neighbors, support vector machines, and many gradient-based models. Decision trees and random forests generally don't require feature scaling.

### Method A — Min-max normalization
`MinMaxScaler` typically maps each feature into the range 0 to 1.

The formula is:

$$x'=\frac{x-x_{\min}}{x_{\max}-x_{\min}}$$

For example, if the minimum study hours is 2, the maximum is 10, and a student studies 6 hours:

$$x'=\frac{6-2}{10-2}=0.5$$

**Code**

````python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

hours = df_cleaned[["study_hours"]]

scaled_hours = scaler.fit_transform(hours)

print(scaled_hours)
````

### Method B — Standardization
`StandardScaler` transforms a feature so that its training data has mean approximately 0 and standard deviation approximately 1.

$$z=\frac{x-\mu}{\sigma}$$

Here, $\mu$ is the mean and $\sigma$ is the standard deviation of the feature in the fitted data.

**Code**

````python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

standardized_hours = scaler.fit_transform(
    df_cleaned[["study_hours"]]
)

print(standardized_hours)
````

**Which one should you use?**
| Property | MinMaxScaler | StandardScaler |
| :--- | :--- | :--- |
| Typical result | Values between 0 and 1 | Mean 0, standard deviation 1 |
| Uses | Minimum and maximum | Mean and standard deviation |
| Sensitive to outliers | Yes | Yes |
| Common applications | Distance-based methods, neural networks | SVM, linear models with regularization, many other estimators |

These are common choices, not rigid rules. The right scaling method depends on the data and algorithm.

<span style="color: #FF3333; font-weight: bold; font-size: 20px">Very important:</span><br>
fit the scaler on training data only. Transform both training and testing data using that fitted scaler. Never calculate the scaling parameters from the complete dataset before splitting.

## 6. Detecting and handling outliers
An outlier is an observation that is unusually far from the rest of the data. It may be a data-entry mistake, a measurement error, or a genuine rare observation.

One common detection method uses the interquartile range (IQR).

$$IQR=Q_3-Q_1
$$

Where:
- $Q_1$ is the 25th percentile.
- $Q_3$ is the 75th percentile.

A common rule flags values below $Q_1-1.5(IQR)$ or above $Q_3+1.5(IQR)$.

**Code**

````python
column = df_cleaned["marks"]

Q1 = column.quantile(0.25)
Q3 = column.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df_cleaned[
    (column < lower_bound) | (column > upper_bound)
]

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
display(outliers)
````

This identifies observations outside the IQR-based bounds. It does not prove they are errors.

**What should we do with outliers?**
1. **Verify them:** check whether the values are valid.
2. **Correct errors:** if a value was recorded incorrectly and the true value is known.
3. **Keep genuine observations:** if the rare values are meaningful.
4. **Consider transformations or robust methods:** when extreme values affect the model in undesirable ways.

Don't automatically delete every outlier. For example, a student scoring 100 marks may be unusual but perfectly valid.

## 7. Building a preprocessing pipeline
So far, we've learned several preprocessing techniques individually. In a real ML project, we often combine them into a single workflow.

Scikit-learn's `Pipeline` lets us chain transformations and a model together. This helps keep training and prediction consistent.

We'll use a slightly more complete dataset for this example so the preprocessing steps are easy to see.

**Explanation**\
Our features will be:
- `study_hours` — numerical
- `attendance` — numerical with missing values
- `department` — categorical with missing values

Our target will be `marks`.

The pipeline will:
1. Impute missing numerical values using the median.
2. Standardize numerical features.
3. Impute missing categorical values using the most frequent category.
4. One-hot encode the categorical feature.
5. Train a linear regression model.

**Code**

````python
from sklearn.model_selection import train_test_split as tts
from sklearn.compose import ColumnTransformer as ct
from sklearn.pipeline import Pipeline as pipe
from sklearn.impute import SimpleImputer as si
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LinearRegression as lr
from sklearn.metrics import mean_absolute_error as mae

# Use the original dataset without the added duplicate
data = df.drop_duplicates().copy()

X = data[["study_hours", "attendance", "department"]]
y = data["marks"]

X_train, X_test, y_train, y_test = tts(
    X,
    y,
    test_size=0.25,
    random_state=45
)

numeric_features = ["study_hours", "attendance"]
categorical_features = ["department"]

numeric_pipeline = pipe([
    ("imputer", si(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = pipe([
    ("imputer", si(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ct([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])

model_pipeline = pipe([
    ("preprocessor", preprocessor),
    ("model", lr())
])

model_pipeline.fit(X_train, y_train)

predictions = model_pipeline.predict(X_test)

print("Predicted marks:", predictions.round(2))
print("Actual marks:", y_test.to_list())
print("Test MAE:", mae(y_test, predictions))
````

**What makes this pipeline useful?**
- The imputers and scaler are fitted only on the training data.
- The encoder learns its categories from the training data.
- The same transformations are automatically applied to test data and future predictions.
- You can call `model_pipeline.predict()` directly on new data in the original feature format.

For example:

````python
new_student = pd.DataFrame({
    "study_hours": [7.5],
    "attendance": [88],
    "department": ["CSE"]
})

predicted_marks = model_pipeline.predict(new_student)

print("Predicted marks:", predicted_marks[0].round(2))
````

The result is a model estimate, not a guarantee of the student's actual score. Our dataset is deliberately tiny and intended for learning.

One more important point: with only eight unique rows, the test set is very small. The purpose here is to learn the preprocessing workflow, not to build a reliable marks-prediction system.

**differences**
1. Range
MinMax is bounded: the smallest value is exactly 0 and the largest exactly 1.\
Standard has no fixed bounds. Here it runs from about −1.73 to +1.90, and values can be any size.

2. Meaning of the numbers
MinMax: 0.43 means “43% of the way from the min to the max.”\
Standard: −0.17 means “0.17 standard deviations below the mean.” Zero is the mean, so negative means below average and positive means above.

3. Center
MinMax output is not centered. The mean here is about 0.48, and it depends on the data.\
Standard output always has mean 0 and standard deviation 1.

4. Sensitivity to outliers
MinMax depends entirely on min and max. One extreme outlier would squash all the other values into a tiny sub-range near 0.\
Standard is also affected, since outliers inflate the mean and std, but it doesn’t squeeze everything into a fixed interval.

5. Shape
Both are linear transformations, so neither changes the shape of the distribution. They only shift and stretch it. The two columns are perfectly correlated.

Use the IQR method to inspect `previous_score` and `final_score`. Explain whether any unusual observations should be removed.

## Task 6 — Pipeline
Create a Scikit-learn pipeline that imputes missing values, scales numerical features, encodes categorical features, and trains a `LinearRegression` model. Split the data before fitting the pipeline and evaluate it using MAE.

````python
from sklearn.model_selection import train_test_split as tts
from sklearn.compose import ColumnTransformer as ct
from sklearn.impute import SimpleImputer as si
from sklearn.pipeline import Pipeline as pipe
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LinearRegression as lr
from sklearn.metrics import mean_absolute_error as mae

# Remove duplicates, but preserve missing values
data = df.drop_duplicates().copy()

X = data[
    ["study_hours", "attendance",
     "department", "previous_score"]
]

y = data["final_score"]

X_train, X_test, y_train, y_test = tts(
    X,
    y,
    test_size=0.25,
    random_state=35
)

numeric_features = ["study_hours", "attendance", "previous_score"]
categorical_features = ["department"]

numeric_pipeline = pipe([
    ("imputer", si(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = pipe([
    ("imputer", si(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ct([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])

model_pipeline = pipe([
    ("preprocessor", preprocessor),
    ("model", lr())
])

model_pipeline.fit(X_train, y_train)

predictions = model_pipeline.predict(X_test)

print("Predicted marks:", predictions)
print("Actual marks:", y_test.to_list())
print("Test MAE:", mae(y_test, predictions))
````

Here we used `df` and not `df_cleaned` or `encoded_df` because in `df_cleaned` we filled the categories before splitting the data and in `encoded_df` we already has the `department` column converted into multiple numerical columns using `pd.get_dummies()`.

In the pipeline, we used Scikit-learn's `OneHotEncoder` to encode department automatically.

If we used `encoded_df`, we would have encoded the `department` twice, or our pipeline wouldn't have find the original `department` column.

**Recommended**\
For the pipeline, start with the original df, remove the accidental duplicate, but leave missing values untouched. Let the pipeline handle imputation after the split.

**1. Understanding the results**
| Study target | Actual marks | Predicted marks | Absolute error |
| :--- | :--- | :--- | :--- |
| Student 1 | 63 | 63.75 | 0.75 |
| Student 2 | 57 | 55.90 | 1.10 |
| Student 3 | 72 | 73.94 | 1.94 |

The model's mean absolute error is approximately 1.26 marks.

$$MAE=\frac{0.75+1.10+1.94}{3}\approx1.26$$

This means the model's predictions differed from the actual scores by an average of approximately 1.26 marks on these three test samples.

**2. What this tells us**
- **The pipeline works:** the missing-value imputers, scaler, encoder, and regression model all ran together.
- **The predictions are reasonably close:** all three predictions are within 2 marks of the actual scores.
- **The test set is very small:** with only three test samples, this MAE is not a reliable estimate of performance on new students generally.

A low MAE on a tiny dataset doesn't necessarily mean the model is excellent. We would need more data and a more reliable evaluation method to draw that conclusion.
