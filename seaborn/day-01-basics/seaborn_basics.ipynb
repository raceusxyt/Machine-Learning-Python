## 1. Introduction to Seaborn

**Seaborn** is a Python data visualization library built on top of Matplotlib. It provides a high-level interface for creating attractive and informative statistical graphics.

### Why use Seaborn?
- Creates visually appealing plots with less code.
- Works smoothly with Pandas DataFrames.
- Provides statistical plots such as heatmaps, box plots, and violin plots.
- Offers built-in datasets for practice.
- Supports themes and color palettes for easy customization.

### Seaborn vs Matplotlib
- **Matplotlib:** Offers detailed control over plot elements.
- **Seaborn:** Makes statistical visualization easier with simpler syntax and attractive defaults.

Seaborn uses Matplotlib behind the scenes, so we can combine both libraries when needed.

## 2. Exploring a Built-in Dataset

Seaborn provides built-in datasets that we can use to practice data visualization without creating a dataset from scratch.

We will use the `tips` dataset, which contains information about restaurant bills and tips.

```python
df = sns.load_dataset("tips")

df.head()
```

## 3. First Visualization — Scatter Plot

A **scatter plot** displays the relationship between two numerical variables using individual points.

We will use:
- **X-axis:** `total_bill`
- **Y-axis:** `tip`

This will help us explore whether customers tend to leave larger tips when their total bill is higher.

```python
sns.scatterplot(data=df, x="total_bill", y="tip")

plt.show()
```

### Observations

- Each point represents one restaurant bill and its corresponding tip.
- The X-axis represents `total_bill`.
- The Y-axis represents `tip`.
- The points generally show a positive relationship: larger bills tend to have larger tips.
- The relationship is not perfect because customers have different tipping habits.

### Customize the plot with hue
Seaborn makes it easy to add a third variable using hue. Let's color the points according to whether the customer was a smoker.

```python
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="smoker"
)

plt.title("Total Bill vs Tip by Smoking Status")

plt.show()
```

### Customize colors and point size
Let's learn three useful parameters:
- hue — colors points by a category.
- palette — controls the colors used for categories.
- s — controls the marker size.

```python
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="smoker",
    palette="Set1",
    s=100
)

plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")
plt.grid(alpha=0.3)

plt.show()
```

Prediction:
1. What effect will palette="Set1" have?
2. What happens when we change s=100 to s=30?
3. Are plt.title() and plt.xlabel() Seaborn functions or Matplotlib functions?

Answer:
1. palette="Set1" — selects the color palette used for the smoker categories.
2. s=100 — sets the scatter markers' size. Changing it to s=30 makes them smaller.
3. plt.title() and plt.xlabel() — these are Matplotlib functions. Seaborn works alongside Matplotlib, so you can use both libraries in the same plot.

### Use style to distinguish categories
Seaborn can distinguish groups using marker shapes as well as colors.

```python
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="sex",
    style="sex",
    s=100
)

plt.title("Total Bill vs Tip by Sex")

plt.show()
```

Prediction:
1. What will hue="sex" change?
2. What will style="sex" change?
3. Why might using both color and marker shape be useful?

Answer:
1. hue="sex" — assigns different colors to the categories in the sex column.
2. style="sex" — assigns different marker shapes to the categories.
3. Using both — makes groups easier to distinguish because they differ in both color and shape.
