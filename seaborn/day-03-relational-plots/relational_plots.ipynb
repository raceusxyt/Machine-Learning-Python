# Imports and Dataset

````python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = sns.load_dataset("tips")

df.head()
````

# Scatter plots
A scatter plot represents the relationship between two numerical variables. Each point represents one observation.

For our dataset:
- `total_bill` — total restaurant bill.
- `tip` — tip amount.
- Each point represents one dining bill and its corresponding tip.

````python
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="smoker"
)

plt.show()
````

You have already seen it in day-01-basics

# Line plots
A line plot is useful for visualizing how a numerical variable changes across an ordered variable, such as time, date, or a sequence.

For a first example, we'll use the tips dataset and examine the average tip by party size.

````python
sns.lineplot(
    data=df,
    x="size",
    y="tip"
)

plt.title("Average Tip by Party Size")
plt.xlabel("Party Size")
plt.ylabel("Average Tip ($)")

plt.show()
````

**Important:** When multiple observations share the same x-value, `lineplot()` aggregates them by default using the mean and displays an uncertainty interval.

**What to observe**
- X-axis: Party size (number of people at the table).
- Y-axis: Average tip amount.
- Line: Shows how the average tip changes with party size.
- Shaded region: Represents the uncertainty interval around the estimate, by default a 95% confidence interval.

Remember: this graph shows the average tip, not each individual customer's tip.

**Important Question:**
If you want to display the individual tip amounts without averaging multiple customers at the same party size, would `sns.lineplot()` be your first choice, or would a scatter plot be more suitable?

**Answer:**
A scatter plot is more suitable for displaying individual customers' tip amounts because each point represents one observation.

## Line plot with hue
Let's extend what you've learned by comparing average tips for smokers and non-smokers.

````python
sns.lineplot(
    data=df,
    x="size",
    y="tip",
    hue="smoker"
)

plt.title("Average Tip by Party Size and Smoking")

plt.show()
````

- Two colored lines: one for smokers and one for non-smokers.
- X-axis: Party size.
- Y-axis: Average tip.
- Legend: Identifies which line represents each smoking group.

## Line plot with style

````python
sns.lineplot(
    data=df,
    x="size",
    y="tip",
    hue="smoker",
    style="smoker"
)

plt.title("Average Tip by Party Size and Smoking")

plt.show()
````

`style="smoker"` make the line styles different

# Relational plots
You've already learned scatterplot() and hue. Now we'll learn something new: faceting, which creates separate plots for different groups.

````python
sns.relplot(
    data=df,
    x="total_bill",
    y="tip",
    col="time",
    kind="scatter"
)

plt.show()
````

`sns.relplot()` creates separate scatter plots for Lunch and Dinner because you used `col="time"`.

What you'll see
- Left panel: Lunch — total bill versus tip.
- Right panel: Dinner — total bill versus tip.
- Both panels use the same variables, making it easier to compare the relationships between lunch and dinner.

Key concept: hue vs. col
| Parameter | What it does |
| :--- | :--- |
| hue="time" | One plot, different colors for Lunch and Dinner |
| col="time" | Separate plots arranged horizontally |
| row="time" | Separate plots arranged vertically |

````python
sns.relplot(
    data=df,
    x="total_bill",
    y="tip",
    row="sex",
    col="time",
    kind="scatter"
)

plt.show()
````

Here:
- row="sex" creates 2 rows: Male and Female.
- col="time" creates 2 columns: Lunch and Dinner.

So the total number of panels is:
$$2 \times 2 = 4 \text{ panels}$$
The layout will look approximately like this:
|	| Lunch | Dinner |
| :--- | :--- | :--- |
| Male | Plot 1 | Plot 2 |
| Female | Plot 3 | Plot 4 |
