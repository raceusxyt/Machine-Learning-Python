### 2. Code
**Import libraries and load data**

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = sns.load_dataset("tips")

df.head()
````

**Apply a theme and create a plot**

````python
sns.set_theme(style="whitegrid", context="notebook")

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="time",
    s=80
)

plt.title("Total Bill vs Tip", fontsize=16)
plt.xlabel("Total Bill ($)", fontsize=12)
plt.ylabel("Tip ($)", fontsize=12)

plt.tight_layout()

plt.show()
````

### 3. Output
What to observe
- A white background with subtle gridlines.
- Different point colors for Lunch and Dinner.
- Larger points because of s=80.
- A title and labeled axes.
- A legend identifying the meal-time categories.

`palette` vs `color`
- `palette` selects a collection of colors, often for different categories.
- `color` sets one color for all plotted elements that support it.

**For example:**
```python
sns.scatterplot(data=df, x="total_bill", y="tip", hue="time",                palette="deep")
```
This assigns colors to the categories in time.
```python
sns.scatterplot(data=df, x="total_bill", y="tip", color="purple")
```
This uses one purple color for all points.

### 2. Code
Experiment with palettes

````python
# Set the theme
sns.set_theme(style="whitegrid", context="talk")

fig, axes = plt.subplots(2, 2, figsize=(13, 10))

fig.suptitle("Color Palettes", fontsize=30, fontweight="bold", color="deeppink")

# Compare two palettes
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="time",
    palette="deep",
    s=70,
    ax=axes[0, 0]
)

axes[0, 0].set_title("Deep Palette")

sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="time",
    palette="bright",
    s=70,
    ax=axes[0, 1]
)

axes[0, 1].set_title("Bright Palette")

sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="time",
    palette="viridis",
    s=70,
    ax=axes[1, 0]
)

axes[1, 0].set_title("Viridis Palette")

sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="time",
    palette="coolwarm",
    s=70,
    ax=axes[1, 1]
)

axes[1, 1].set_title("Coolwarm Palette")

plt.tight_layout()

plt.show()
````

## Lesson 3: Figure Size, Titles, Labels, and Layout
### 1. Explanation
**A. Figure size — `figsize`**
Controls the overall width and height of a Matplotlib figure, measured in inches.
```python
plt.figure(figsize=(10, 5))
```
- 10 = width
- 5 = height

**B. Titles and labels**
| Function | Purpose |
| :--- | :--- |
| `plt.title()` | Adds a title to a plot |
| `plt.xlabel()` | Labels the x-axis |
| `plt.ylabel()` | Labels the y-axis |
| `plt.suptitle()` | Adds a title above an entire figure|

You can also use `ax.set_title()`, `ax.set_xlabel()`, and `ax.set_ylabel()` when working with subplot axes.

**C. Font customization**
```python
plt.title("Daily Expenses", fontsize=16, fontweight="bold")
plt.xlabel("Date", fontsize=12)
plt.ylabel("Amount", fontsize=12)
```
**D. Tick labels and layout**
```python
plt.xticks(rotation=45)
plt.tight_layout()
```
- `rotation=45` rotates x-axis tick labels by 45 degrees, making long labels easier to read.
- `tight_layout()` adjusts spacing to reduce overlaps between plot elements.

### 2. Code: Single plot

````python
# Set the theme
sns.set_theme(style="darkgrid", context="talk")

# Customize a scatter plot
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="time",
    palette="deep",
    s=70
)

plt.title(
    "Relationship Between Total Bill and Tip",
    fontsize="30",
    fontweight="bold",
)
plt.xlabel("Total Bill ($)", fontsize=20)
plt.ylabel("Tip ($)", fontsize=20)
plt.xticks(rotation=30)

plt.tight_layout()

plt.show()
````

### 3. Output: Single plot
**What to observe:**
- The plot has a larger canvas
- It has bold title
- Readable axis labels
- Colored categories
- Rotated x-axis ticks.

### 4. Code: Multiple plots — `subplots()`
When you create a dashboard, use `plt.subplots()` to control multiple plots together.

````python
sns.set_theme(style="whitegrid", context="talk")

fig, axes = plt.subplots(1, 2, figsize=(14, 7))

fig.suptitle("Tips Dataset Dashboard", fontsize=30, fontweight="bold", color="deeppink")

sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="time",
    ax=axes[0]
)

axes[0].set_title("Total Bill vs Tip", fontsize=20, fontweight="bold")

sns.boxplot(
    data=df,
    x="day",
    y="total_bill",
    hue="day",
    dodge=False,
    legend=False,
    ax=axes[1]
)

axes[1].set_title("Bill Distribution by Day", fontsize=20, fontweight="bold")

plt.tight_layout()

plt.show()
````

### 5. Output: Multipleplots — `subplots()`
**What to observe**
- `fig` represents the entire figure, while `axes[0]` and `axes[1]` represent the individual plots.
- The second plot uses `hue="day"` to assign colors to days
- `legend=False` hides the redundant legend.

````python
# Set the overall theme
sns.set_theme(style="whitegrid", context="talk")

# Create a two-panel dashboard
fig, axes = plt.subplots(1, 2, figsize=(16, 8))

# Shared dashboard title
fig.suptitle(
    "Seaborn dashboard title",
    fontsize=40,
    fontweight="bold",
    color="deeppink"
)

# Plot 1: Average tip by day
sns.barplot(
    data=df,
    x="day",
    y="tip",
    hue="day",
    palette="magma",
    dodge=False,
    legend=False,
    errorbar=None,
    ax=axes[0]
)

axes[0].set_title(
    "Average Tip by Day",
    fontsize=30,
    fontweight="bold"
)
axes[0].set_xlabel("Day", fontsize=20, fontweight="bold")
axes[0].set_ylabel("Average Tip ($)", fontsize=20, fontweight="bold")

sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="time",
    palette="Set2",
    s=70,
    alpha=0.8,
    ax=axes[1]
)

axes[1].set_title(
    "Total Bill vs Tip",
    fontsize=30,
    fontweight="bold",
)

axes[1].set_xlabel("Total Bill ($)", fontsize=20, fontweight="bold")
axes[1].set_ylabel("Tip ($)", fontsize=20, fontweight="bold")

# Adjust spacing and display
plt.tight_layout()

plt.show()
````

### 4. Step 3 — Understand the important parameters
| Parameter | Purpose |
| :---| :--- |
| `sns.set_theme()` | Sets the overall plot style |
| `figsize=(14, 5)` | Sets figure width and height |
| `palette="viridis"` | Selects colors for the bar categories |
| `palette="Set2"` | Selects colors for Lunch and Dinner
| `dodge=False` | Overrides Seaborn's default behavior of trying to create separate sub-columns for each category. |
| `errorbar=None` |	Hides the bar plot's uncertainty intervals |
| `s=70` | Sets scatter marker size |
| `alpha=0.8` | Controls point transparency |
| `fig.suptitle()` | Adds the overall dashboard title |
| `plt.tight_layout()` | Adjusts spacing between plot elements |

**Important:** `hue="day"` colors the bars by day. Since it duplicates the x-axis categories, `legend=False` avoids displaying a redundant legend.
